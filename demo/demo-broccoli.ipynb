# Diverse desires: Broccoli and Goldfish

**Inspired by:** Repacholi, B. M., & Gopnik, A. (1997). Early reasoning about desires: evidence from 14-and 18-month-olds. Developmental psychology, 33(1), 12.

Repacholi and Gopnik's task shows that children at 18 months can represent that others have different desires from their own, and can infer those desires from their displays of emotion. Here, we show that memo can model this behavior _without_ requiring the child to use the construct `thinks`.

```python
from memo import memo
from typing import no_type_check
from enum import IntEnum
```

```python
class Treat(IntEnum):
    BROCCOLI = 0
    GOLDFISH = 1

class EmotionExpression(IntEnum):
    SMILE = 0
    SCOWL = 1

@no_type_check
@memo
def main[broccoli_ee: EmotionExpression, offered_treat: Treat]():
    child: has_theory(CHILD)
    child: thinks[
        adult: given(preferred_treat in Treat, uniformly),
        adult: chooses(broccoli_ee in EmotionExpression, such_that=
            (broccoli_ee == {EmotionExpression.SMILE})
            if preferred_treat == {Treat.BROCCOLI} else
            (broccoli_ee == {EmotionExpression.SCOWL})
        )
    ]
    child: observes [adult.broccoli_ee] is broccoli_ee
    child: chooses(offered_treat in Treat, to_maximize=Pr[offered_treat == adult.preferred_treat])
    return Pr[child.offered_treat == offered_treat]

main(print_table=True);
```
