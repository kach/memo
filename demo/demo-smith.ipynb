# The Mr. Smith Problem

**Inspired by:** Bar-Hillel, M., & Falk, R. (1982). Some teasers concerning conditional probabilities. Cognition, 11(2), 109-122.

> Mr. Smith has two children. **At least one of them is a boy.** What is the probability that both children are boys?

```python
from memo import memo
from enum import IntEnum
from typing import no_type_check
```

```python
class Gender(IntEnum):
    # Sorry to have to reinforce the binary :(
    BOY = 0
    GIRL = 1

class Child(IntEnum):
    OLDER = 0
    YOUNGER = 1

@no_type_check
@memo
def interpretation_1():
    '''
    Interpret the sentence as: either c1 is a boy, or c2 is a boy, or both.
    '''
    reader: thinks[
        smith: given(c1 in Gender, c2 in Gender, uniformly)
    ]
    reader: observes_that[smith.c1 == {Gender.BOY} or smith.c2 == {Gender.BOY}]
    return reader[Pr[smith.c1 == {Gender.BOY}, smith.c2 == {Gender.BOY}]]
print('Interpretation 1:', interpretation_1())

@no_type_check
@memo
def interpretation_2():
    '''
    Interpret the sentence as: one of the children was observed to be a boy.
    '''
    reader: thinks[
        smith: given(c1 in Gender, c2 in Gender, uniformly),
        smith: chooses(which_child in Child, uniformly),
        smith: chooses(that_child_gender in Gender, to_be=c1 if which_child == 0 else c2)
    ]
    reader: observes_that[smith.that_child_gender == {Gender.BOY}]
    return reader[Pr[smith.c1 == {Gender.BOY}, smith.c2 == {Gender.BOY}]]
print('Interpretation 2:', interpretation_2())
```
